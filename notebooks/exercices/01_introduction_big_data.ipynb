## 5ï¸âƒ£ DÃ©monstration : Limites de Pandas

````python
# Importons Pandas
import pandas as pd
import time

# CrÃ©ons un petit DataFrame
petit_df = pd.DataFrame({
    'id': range(1000),
    'valeur': range(1000)
})

print("Petit DataFrame (1 000 lignes) :")
print(petit_df.head())
print(f"\nTaille en mÃ©moire : {petit_df.memory_usage(deep=True).sum() / 1024:.2f} Ko")
````

````python
# CrÃ©ons un DataFrame plus grand
import numpy as np

# 1 million de lignes
nb_lignes = 1_000_000

debut = time.time()
grand_df = pd.DataFrame({
    'id': range(nb_lignes),
    'valeur_a': np.random.rand(nb_lignes),
    'valeur_b': np.random.rand(nb_lignes),
    'categorie': np.random.choice(['A', 'B', 'C', 'D'], nb_lignes)
})
fin = time.time()

print(f"DataFrame crÃ©Ã© en {fin - debut:.2f} secondes")
print(f"Nombre de lignes : {len(grand_df):,}")
print(f"Taille en mÃ©moire : {grand_df.memory_usage(deep=True).sum() / (1024*1024):.2f} Mo")
````

````python
# AgrÃ©gation sur le grand DataFrame
debut = time.time()
resultat = grand_df.groupby('categorie').agg({
    'valeur_a': 'mean',
    'valeur_b': 'sum'
})
fin = time.time()

print(f"AgrÃ©gation terminÃ©e en {fin - debut:.4f} secondes")
print("\nRÃ©sultat :")
print(resultat)
````

````python
import pandas as pd
import numpy as np
import time

# Creez un DataFrame avec 5 millions de lignes
nb_lignes = 5_000_000  # 5 millions

debut = time.time()

df = pd.DataFrame({
    'id': np.arange(nb_lignes),
    'valeur_a': np.random.rand(nb_lignes),
    'valeur_b': np.random.rand(nb_lignes),
    'categorie': np.random.choice(['A', 'B', 'C', 'D'], nb_lignes)
})

fin = time.time()

taille_mo = df.memory_usage(deep=True).sum() / (1024 * 1024)
debut_agg = time.time()
resultat = df.groupby('categorie').agg({
    'valeur_a': 'mean',
    'valeur_b': 'sum'
})
fin_agg = time.time()

print(f"Temps de creation : {fin - debut:.2f} secondes")
print(f"Taille memoire : {taille_mo:.2f} Mo")
print(f"Temps d'aggregation : {fin_agg - debut_agg:.4f} secondes")
print("\nResultat de l'aggregation :")
print(resultat)
````
